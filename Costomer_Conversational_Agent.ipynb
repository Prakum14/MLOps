```python
def plotPerColumnDistribution(df, nGraphShown, nGraphPerRow):
    nunique = df.nunique()
    df = df[[col for col in df if nunique[col] > 1 and nunique[col] < 50]] # For displaying purposes, pick columns that have between 1 and 50 unique values
    nRow, nCol = df.shape
    columnNames = list(df)
    nGraphRow = int((nCol + nGraphPerRow - 1) / nGraphPerRow)
    plt.figure(num = None, figsize = (6 * nGraphPerRow, 8 * nGraphRow), dpi = 80, facecolor = 'w', edgecolor = 'k')
    for i in range(min(nCol, nGraphShown)):
        plt.subplot(nGraphRow, nGraphPerRow, i + 1)
        columnDf = df.iloc[:, i]
        if (not np.issubdtype(type(columnDf.iloc[0]), np.number)):
            valueCounts = columnDf.value_counts()
            valueCounts.plot.bar()
        else:
            columnDf.hist()
        plt.ylabel('counts')
        plt.xticks(rotation = 90)
        plt.title(f'{columnNames[i]} (column {i})')
    plt.tight_layout(pad = 1.0, w_pad = 1.0, h_pad = 1.0)
    plt.show()
```

```python
def plotCorrelationMatrix(df, graphWidth):
    filename = df.dataframeName
    df = df.dropna('columns') # drop columns with NaN
    df = df[[col for col in df if df[col].nunique() > 1]] # keep columns where there are more than 1 unique values
    if df.shape[1] < 2:
        print(f'No correlation plots shown: The number of non-NaN or constant columns ({df.shape[1]}) is less than 2')
        return
    corr = df.corr()
    plt.figure(num=None, figsize=(graphWidth, graphWidth), dpi=80, facecolor='w', edgecolor='k')
    corrMat = plt.matshow(corr, fignum = 1)
    plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
    plt.yticks(range(len(corr.columns)), corr.columns)
    plt.gca().xaxis.tick_bottom()
    plt.colorbar(corrMat)
    plt.title(f'Correlation Matrix for {filename}', fontsize=15)
    plt.show()
```

```python
def plotScatterMatrix(df, plotSize, textSize):
    df = df.select_dtypes(include =[np.number]) # keep only numerical columns
    # Remove rows and columns that would lead to df being singular
    df = df.dropna('columns')
    df = df[[col for col in df if df[col].nunique() > 1]] # keep columns where there are more than 1 unique values
    columnNames = list(df)
    if len(columnNames) > 10: # reduce the number of columns for matrix inversion of kernel density plots
        columnNames = columnNames[:10]
    df = df[columnNames]
    ax = pd.plotting.scatter_matrix(df, alpha=0.75, figsize=[plotSize, plotSize], diagonal='kde')
    corrs = df.corr().values
    for i, j in zip(*plt.np.triu_indices_from(ax, k = 1)):
        ax[i, j].annotate('Corr. coef = %.3f' % corrs[i, j], (0.8, 0.2), xycoords='axes fraction', ha='center', va='center', size=textSize)
    plt.suptitle('Scatter and Density Plot')
    plt.show()
```

```python
# --- 2. Define a reusable cleaning function ---
def clean_conversation_dataframechat(df, question):
    """
    Performs data cleaning on a pandas DataFrame, specifically designed for
    conversation text columns.

    Steps include:
    1. Removing null values (by filling with empty strings).
    2. Standardizing text (lowercase, removing extra whitespace).
    3. Attempting to correct common encoding issues (mojibake).

    Args:
        df (pd.DataFrame): The DataFrame to be cleaned. A copy is made internally
                           to avoid modifying the original DataFrame directly.
        text_column_name (str): The name of the column in the DataFrame that
                                contains the conversation text.

    Returns:
        pd.DataFrame: A new DataFrame with the cleaned conversation text.
    """
    cleaned_df = df.copy() # Work on a copy to preserve the original DataFrame
    print(f"\n--- Starting cleaning for DataFrame column: '{question}' ---")
    print(f"Initial rows: {len(cleaned_df)}")

    # --- 2.1. Remove Nulls ---
    # For conversational data, it's often better to replace NaN with empty strings
    # instead of dropping rows, to maintain the conversation context.
    # If a message is truly non-existent or irrelevant, it will become an empty string.
    null_count_before = cleaned_df[question].isnull().sum()
    if null_count_before > 0:
        cleaned_df[question] = cleaned_df[question].fillna('')
        print(f"Filled {null_count_before} null values in '{question}' with empty strings.")
    else:
        print(f"No null values found in '{question}'.")

    # --- 2.2. Standardize Text ---
    # Ensure the column is of string type before applying string operations
    cleaned_df[question] = cleaned_df[question].astype(str)

    # Convert all text to lowercase
    cleaned_df[question] = cleaned_df[question].str.lower()
    print("Converted text to lowercase.")

    # Remove leading and trailing whitespace
    cleaned_df[question] = cleaned_df[question].str.strip()
    print("Removed leading/trailing whitespace.")

    # Replace multiple internal spaces with a single space using regex
    # This also handles tabs, newlines, and other whitespace characters
    cleaned_df[question] = cleaned_df[question].apply(
        lambda x: re.sub(r'\s+', ' ', x)
    )
    print("Standardized internal whitespace.")

    # --- 2.3. Correct Encodings (Heuristic approach for common mojibake) ---
    # This attempts to fix common "mojibake" issues where UTF-8 characters
    # were incorrectly decoded as Latin-1 and then re-interpreted.
    # Example: 'â\x80\x93' -> '–' (en dash)
    def fix_mojibake_encoding(text):
        if not isinstance(text, str) or not text: # Skip if not string or empty
            return text
        try:
            # This specific sequence can fix certain common mojibake patterns.
            return text.encode('latin1').decode('utf8')
        except (UnicodeEncodeError, UnicodeDecodeError):
            # If the above fails, try a simpler approach if needed or return original
            # Example: text.replace('â\x80\x99', "'").replace('â\x80\x9c', '"')...
            return text # Return original if heuristic doesn't apply or fails

    cleaned_df[question] = cleaned_df[question].apply(fix_mojibake_encoding)
    print("Attempted to correct common encoding issues (mojibake).")

    print(f"Finished cleaning for DataFrame column: '{question}'.")
    print(f"Final rows: {len(cleaned_df)}")
    return cleaned_df
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
import numpy as np

# --- 1. Re-create or load your cleaned DataFrames (as done in the previous step) ---
# For demonstration purposes, I'll use the dummy cleaned DataFrames from the previous step.
# In a real scenario, you'd use dfchatbotconverse_cleaned, dftwitterchat_cleaned, dfallindustconverse_cleaned
# that resulted from your actual data cleaning.

# --- Dummy DataFrames for demonstration (assuming they are already cleaned) ---
# If you are running this after the previous cleaning script,
# you can comment out these dummy dataframes and use the `_cleaned` ones directly.
#dfchatbotconverse = pd.DataFrame({
   # 'conversation_id': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    #'speaker': ['cust', 'agent'] * 5,
    #'text': [f'chat message {i}' for i in range(10)]
#})
#dftwitterchat = pd.DataFrame({
   # 'tweet_id': [201, 202, 203, 204, 205, 206, 207, 208, 209, 210, 211, 212, 213, 214, 215],
    #'user_handle': ['@userA', '@userB', '@userC'] * 5,
    #'text': [f'tweet {i}' for i in range(15)]
#})
#dfallindustconverse = pd.DataFrame({
    #'session_id': [301, 302, 303, 304, 305, 306, 307, 308, 309, 310, 311, 312, #313, 314, 315, 316, 317, 318, 319, 320],
   # 'participant_type': ['cust', 'support'] * 10,
    #'message_content': [f'industry msg {i}' for i in range(20)]
#})

# Rename for consistency if you're using the direct output from previous cleaning
dfchatbotconverse_cleaned = dfchatbotconverse_cleanedchat
dftwitterchat_cleaned = dftwitterchat_cleaned
dfalldatachat_cleaned = dfalldatachat_cleaned

print("Cleaned DataFrames loaded (or dummy data generated).\n")


# --- 2. Define a function to perform the train/validation/test split ---
def perform_train_val_test_split(df, test_size=0.2, val_size=0.5, random_state=42):
    """
    Splits a DataFrame into training, validation, and test sets.

    Args:
        df (pd.DataFrame): The input DataFrame to split.
        test_size (float): The proportion of the dataset to include in the test split.
                           (e.g., 0.2 for 20% of total data)
        val_size (float): The proportion of the *remaining* data (after test split)
                          to include in the validation split.
                          (e.g., 0.5 for 50% of the remaining 20% will be validation,
                          making it 10% of the total data)
        random_state (int): Controls the shuffling applied to the data before splitting.
                            Set for reproducibility.

    Returns:
        tuple: (train_df, val_df, test_df) - The split DataFrames.
    """
    print(f"  Splitting DataFrame of shape: {df.shape}")

    # Step 1: Split data into training set and a temporary set (for validation + test)
    # The 'test_size' here defines the proportion of the *original* DataFrame
    train_df, temp_df = train_test_split(df, test_size=test_size, random_state=random_state)
    print(f"  Train set initial shape: {train_df.shape}")
    print(f"  Temp set (val+test) initial shape: {temp_df.shape}")

    # Step 2: Split the temporary set into validation and test sets
    # The 'test_size' here defines the proportion of the *temp_df*
    val_df, test_df = train_test_split(temp_df, test_size=val_size, random_state=random_state)
    print(f"  Validation set shape: {val_df.shape}")
    print(f"  Test set shape: {test_df.shape}")

    # Verify proportions
    total_rows = len(df)
    print(f"  Approximate splits: Train={len(train_df)/total_rows:.2%}, Val={len(val_df)/total_rows:.2%}, Test={len(test_df)/total_rows:.2%}")

    return train_df, val_df, test_df


# --- 3. Apply the split function to each cleaned DataFrame ---

print("--- Splitting dfchatbotconverse_cleaned ---")
chatbot_train, chatbot_val, chatbot_test = perform_train_val_test_split(
    dfchatbotconverse_cleaned,
    test_size=0.2, # 20% for temp_df (val+test)
    val_size=0.5,  # 50% of temp_df goes to val, 50% to test -> 10% each of original
    random_state=42
)

print("\n--- Splitting dftwitterchat_cleaned ---")
twitter_train, twitter_val, twitter_test = perform_train_val_test_split(
    dftwitterchat_cleaned,
    test_size=0.2,
    val_size=0.5,
    random_state=42
)

print("\n--- Splitting dfallindustconverse_cleaned ---")
allindust_train, allindust_val, allindust_test = perform_train_val_test_split(
    dfalldatachat_cleaned,
    test_size=0.2,
    val_size=0.5,
    random_state=42
)


# --- 4. Print the shapes of the resulting datasets to confirm ---
print("\n" + "="*50)
print("--- Final Shapes of Split DataFrames ---")
print(f"dfchatbotconverse_cleaned:")
print(f"  Train: {chatbot_train.shape}")
print(f"  Validation: {chatbot_val.shape}")
print(f"  Test: {chatbot_test.shape}")

print(f"\ndftwitterchat_cleaned:")
print(f"  Train: {twitter_train.shape}")
print(f"  Validation: {twitter_val.shape}")
print(f"  Test: {twitter_test.shape}")

print(f"\ndfallindustconverse_cleaned:")
print(f"  Train: {allindust_train.shape}")
print(f"  Validation: {allindust_val.shape}")
print(f"  Test: {allindust_test.shape}")
print("="*50 + "\n")

# Example: Display head of one of the split dataframes
print("\n--- Sample of chatbot_train data ---")
print(chatbot_train.head())
print("\n--- Sample of twitterchat_train data ---")
print(twitter_train.head())
print("\n--- Sample of allindust_train data ---")
print(allindust_train.head())
```
